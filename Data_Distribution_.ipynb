1. Normal Distribution (Gaussian Distribution)
⭐ What it is:

A symmetric bell-shaped curve where values are most concentrated around the mean.
It is the most common distribution found in nature.

⭐ Key Features:

Mean = median = mode

68% of data within 1 standard deviation

95% of data within 2 standard deviations

Perfect for modeling “natural variation”

⭐ Real-World Examples:

Human heights (most are around average; few very short/tall)

IQ scores (most people score near the mean)

Measurement errors in scientific instruments

Blood pressure readings in a population

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(0)
normal_data = np.random.normal(loc=0, scale=1, size=1000)

plt.figure(figsize=(8,5))
sns.histplot(normal_data, kde=True, bins=30)
plt.title("Normal Distribution (mean=0, std=1)")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

2. Binomial Distribution
⭐ What it is:

Used when there are only two outcomes (success / failure) repeated multiple times.

⭐ Key Conditions:

Each trial has two outcomes

Fixed number of trials

Probability remains constant

⭐ Real-World Examples:

Tossing a coin 20 times and counting number of heads

Quality control: out of 100 items, how many defective?

Customer behavior: Will a customer buy or not buy?

MCQ exam: Probability of guessing correct answers

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(0)
binomial_data = np.random.binomial(n=10, p=0.5, size=1000)

plt.figure(figsize=(8,5))
sns.histplot(binomial_data, kde=False, bins=10)
plt.title("Binomial Distribution (n=10, p=0.5)")
plt.xlabel("Number of Successes")
plt.ylabel("Frequency")
plt.show()
```

3. Poisson Distribution
⭐ What it is:

Models the number of rare events happening in a fixed time, space, or distance.

⭐ Key Idea:

Used when events happen randomly and independently.

⭐ Real-World Examples:

Number of customers arriving at a shop per hour

Number of emails you get in a day

Number of accidents on a highway per month

Call center: Number of calls arriving per minute

Hospital emergency room: Patient arrivals per hour

```python
np.random.seed(0)
poisson_data = np.random.poisson(lam=4, size=1000)

plt.figure(figsize=(8,5))
sns.histplot(poisson_data, kde=False, bins=15)
plt.title("Poisson Distribution (lambda=4)")
plt.xlabel("Events")
plt.ylabel("Frequency")
plt.show()
```

4. Uniform Distribution
⭐ What it is:

Every value in the range has an equal probability.

⭐ Two Types:

Discrete Uniform → integers

Continuous Uniform → fractional values

⭐ Real-World Examples:

Rolling a fair die (1 to 6 equal chance)

Random password generator choosing characters

Shuffling cards — each card is equally likely

Random number between 0 and 1 for simulation

```python
np.random.seed(0)
uniform_data = np.random.uniform(low=0, high=1, size=1000)

plt.figure(figsize=(8,5))
sns.histplot(uniform_data, kde=True, bins=30)
plt.title("Uniform Distribution (0 to 1)")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()

```

5. Logistic Distribution
⭐ What it is:

Similar to normal distribution but with heavier tails, meaning extreme values are more likely.
Used heavily in machine learning & growth modeling.

⭐ Real-World Examples:

Logistic regression: predicting if a customer will buy (0/1)

Population growth: growth starts slow → increases → slows again

Epidemics: spread of a virus (infection probability curve)

Adoption of technology: slow → rapid → saturation

```python
np.random.seed(0)
logistic_data = np.random.logistic(loc=0, scale=1, size=1000)

plt.figure(figsize=(8,5))
sns.histplot(logistic_data, kde=True, bins=30)
plt.title("Logistic Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

6. Multinomial Distribution
⭐ What it is:

A generalization of binomial distribution to more than two outcomes.

⭐ Example Cases:**

Instead of just “success/failure”, you have multiple categories.

⭐ Real-World Examples:

Rolling a dice 100 times → count how many 1s, 2s, 3s…

Survey responses: “Yes / No / Maybe”

Customer choices: pizza / burger / pasta

Voting results: votes for different candidates

```python
np.random.seed(0)
outcomes = np.random.multinomial(100, [0.2, 0.3, 0.5], size=1)[0]

plt.figure(figsize=(8,5))
sns.barplot(x=["Category A", "Category B", "Category C"], y=outcomes)
plt.title("Multinomial Distribution")
plt.xlabel("Categories")
plt.ylabel("Counts")
plt.show()
```

7. Exponential Distribution
⭐ What it is:

Models the time between events in a Poisson process.
Often called the waiting time distribution.

⭐ Real-World Examples:

Time between bus arrivals

Time before a machine part breaks

Time before next customer arrives at a shop

Time to next earthquake, aftershocks

Duration of phone calls / waiting time

```python
np.random.seed(0)
exponential_data = np.random.exponential(scale=1, size=1000)

plt.figure(figsize=(8,5))
sns.histplot(exponential_data, kde=True, bins=30)
plt.title("Exponential Distribution (scale=1)")
plt.xlabel("Time between events")
plt.ylabel("Frequency")
plt.show()
```

8. Chi-Square Distribution
⭐ What it is:

Used mainly in hypothesis testing, especially with categorical data.
Measures how much observed data differs from expected data.

⭐ Real-World Examples:

Testing if a dice is fair

Testing relationship between gender & product preference

Goodness of fit tests for distribution patterns

Genetics: testing inheritance patterns

Association tests: education vs income category

```python
np.random.seed(0)
chisquare_data = np.random.chisquare(df=4, size=1000)

plt.figure(figsize=(8,5))
sns.histplot(chisquare_data, kde=True, bins=30)
plt.title("Chi-Square Distribution (df=4)")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```
